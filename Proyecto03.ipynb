```python
import os
from reportlab.lib.pagesizes import letter, A4
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib.units import inch
from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer, Table, TableStyle, PageBreak, Image
from reportlab.lib import colors
from datetime import datetime
import io

# Crear el PDF
pdf_filename = "Informe_Proyecto3.pdf"
doc = SimpleDocTemplate(pdf_filename, pagesize=letter, topMargin=0.5*inch, bottomMargin=0.5*inch)
story = []

# Estilos
styles = getSampleStyleSheet()
title_style = ParagraphStyle(
    'CustomTitle',
    parent=styles['Heading1'],
    fontSize=24,
    textColor=colors.HexColor('#1f77b4'),
    spaceAfter=30,
    alignment=1  # Center
)

heading_style = ParagraphStyle(
    'CustomHeading',
    parent=styles['Heading2'],
    fontSize=14,
    textColor=colors.HexColor('#1f77b4'),
    spaceAfter=12,
    spaceBefore=12
)

# Título
story.append(Paragraph("INFORME DE ANÁLISIS DE ACCIONES", title_style))
story.append(Paragraph(f"Generado: {datetime.now().strftime('%d/%m/%Y %H:%M')}", styles['Normal']))
story.append(Spacer(1, 0.3*inch))

# Resumen del proyecto
story.append(Paragraph("1. DESCRIPCIÓN DEL PROYECTO", heading_style))
descripcion = """
Este proyecto realiza un análisis integral de tres acciones tecnológicas principales (Apple, 
Nvidia y Microsoft) durante el período de un año. El análisis incluye estadísticas descriptivas, 
visualizaciones interactivas y un informe generado por Inteligencia Artificial.
"""
story.append(Paragraph(descripcion, styles['BodyText']))
story.append(Spacer(1, 0.2*inch))

# Estadísticas
story.append(Paragraph("2. ESTADÍSTICAS DESCRIPTIVAS", heading_style))
story.append(Paragraph("Resumen estadístico de los precios de cierre:", styles['Normal']))
story.append(Spacer(1, 0.1*inch))

# Crear tabla con las estadísticas
stats_data = [['Métrica', 'AAPL', 'NVDA', 'MSFT']]
stats_metrics = precios.describe()

for metric in stats_metrics.index:
    row = [metric]
    for ticker in ['AAPL', 'NVDA', 'MSFT']:
        value = stats_metrics.loc[metric, ticker]
        row.append(f"${value:.2f}")
    stats_data.append(row)

stats_table = Table(stats_data, colWidths=[1.5*inch, 1.2*inch, 1.2*inch, 1.2*inch])
stats_table.setStyle(TableStyle([
    ('BACKGROUND', (0, 0), (-1, 0), colors.HexColor('#1f77b4')),
    ('TEXTCOLOR', (0, 0), (-1, 0), colors.whitesmoke),
    ('ALIGN', (0, 0), (-1, -1), 'CENTER'),
    ('FONTNAME', (0, 0), (-1, 0), 'Helvetica-Bold'),
    ('FONTSIZE', (0, 0), (-1, 0), 10),
    ('BOTTOMPADDING', (0, 0), (-1, 0), 12),
    ('GRID', (0, 0), (-1, -1), 1, colors.black),
    ('ROWBACKGROUNDS', (0, 1), (-1, -1), [colors.white, colors.HexColor('#f0f0f0')])
]))

story.append(stats_table)
story.append(Spacer(1, 0.2*inch))

# Análisis de IA
story.append(Paragraph("3. ANÁLISIS DE INTELIGENCIA ARTIFICIAL", heading_style))
try:
    # Intentar usar el análisis de IA si está disponible
    story.append(Paragraph("Informe generado por Google Gemini AI:", styles['Normal']))
    story.append(Spacer(1, 0.1*inch))
    story.append(Paragraph(respuesta.text, styles['BodyText']))
except NameError:
    # Si respuesta no está definida, crear un análisis manual
    story.append(Paragraph("Análisis Manual de las Acciones:", styles['Normal']))
    story.append(Spacer(1, 0.1*inch))
    
    analysis_text = "<b>Resumen por Acción:</b><br/><br/>"
    for ticker in ['AAPL', 'NVDA', 'MSFT']:
        final_price = precios[ticker].iloc[-1]
        initial_price = precios[ticker].iloc[0]
        max_price = precios[ticker].max()
        min_price = precios[ticker].min()
        avg_price = precios[ticker].mean()
        change_percent = ((final_price - initial_price) / initial_price) * 100
        
        analysis_text += f"""
<b>{ticker}:</b> El precio inicial fue ${initial_price:.2f} y el precio final fue ${final_price:.2f}. 
El cambio porcentual fue de {change_percent:+.2f}%. El precio máximo alcanzado fue ${max_price:.2f}, 
el mínimo fue ${min_price:.2f}, y el promedio del año fue ${avg_price:.2f}.<br/><br/>
"""
    
    story.append(Paragraph(analysis_text, styles['BodyText']))

story.append(Spacer(1, 0.2*inch))

# Gráficos generados
story.append(Paragraph("4. GRÁFICOS GENERADOS", heading_style))
graphics_info = """
<b>Se han generado los siguientes gráficos interactivos:</b><br/>
• Gráfico de violín: Distribución de precios de acciones<br/>
• Gráfico de línea: Evolución histórica de precios<br/>
• Gráfico de línea: Precio promedio de las tres acciones<br/>
• Gráfico de área: Distribución del volumen<br/>
• Gráfico de dispersión: Evolución de inversión de $1000<br/>
• Gráfico de línea principal: Guardado como <b>inversion_1000_dolares.html</b><br/>
"""
story.append(Paragraph(graphics_info, styles['BodyText']))
story.append(Spacer(1, 0.2*inch))

# Conclusiones
story.append(Paragraph("5. DATOS FINALES", heading_style))
final_data = []
for ticker in ['AAPL', 'NVDA', 'MSFT']:
    final_price = precios[ticker].iloc[-1]
    initial_price = precios[ticker].iloc[0]
    change_percent = ((final_price - initial_price) / initial_price) * 100
    final_data.append([ticker, f"${initial_price:.2f}", f"${final_price:.2f}", f"{change_percent:+.2f}%"])

final_table_data = [['Acción', 'Precio Inicial', 'Precio Final', 'Variación %']] + final_data
final_table = Table(final_table_data, colWidths=[1.2*inch, 1.2*inch, 1.2*inch, 1.2*inch])
final_table.setStyle(TableStyle([
    ('BACKGROUND', (0, 0), (-1, 0), colors.HexColor('#1f77b4')),
    ('TEXTCOLOR', (0, 0), (-1, 0), colors.whitesmoke),
    ('ALIGN', (0, 0), (-1, -1), 'CENTER'),
    ('FONTNAME', (0, 0), (-1, 0), 'Helvetica-Bold'),
    ('FONTSIZE', (0, 0), (-1, 0), 10),
    ('BOTTOMPADDING', (0, 0), (-1, 0), 12),
    ('GRID', (0, 0), (-1, -1), 1, colors.black),
    ('ROWBACKGROUNDS', (0, 1), (-1, -1), [colors.white, colors.HexColor('#f0f0f0')])
]))

story.append(final_table)
story.append(Spacer(1, 0.3*inch))

# Pie de página
footer_text = f"<i>Este informe fue generado automáticamente por el Sistema de Análisis de Acciones</i>"
story.append(Paragraph(footer_text, styles['Normal']))

# Generar PDF
doc.build(story)
print(f"✅ PDF generado exitosamente: {pdf_filename}")
print(f"📊 Ruta del archivo: {pdf_filename}")
print(f"📁 El archivo se encuentra en: {os.getcwd()}/{pdf_filename}")
```
