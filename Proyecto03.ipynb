```python
import plotly.express as px 
import plotly.io as pio
```

```python
px.violin(precios,#["AAPL"]este es un filtro que muestra solo apple, si queremos ver todas
        title="Precio de la acción de Apple en el último año",
        labels={"value":"Precio de la acción de Apple (USD)", "Date":"Fecha"},
        template="plotly_dark"#Grafica oscura, si queremos una grafica clara podemos pon
        
)#llamamos a la funcion line de plotly express y le pasamos como parametro la co
```

```python
import plotly.io as pio
#print(pio.renderers) # imprimimos el renderizador por defecto de plotly express, que es 
#pio.renderers.default = "vscode" # cambiamos el renderizador por defecto de plotly expre
inversion = (precios / precios.values[0]) * 1000
px.scatter(inversion,
        title="Evolución de una inversión de $1000 en cada acción",#calculamos la evolucio
        labels={"value":"Valor de la inversión (USD)", "Date":"Fecha"},
        template="plotly_dark"#Grafica oscura, si queremos una grafica clara podemos poner
        )
```

#guardar grafica

```python
fig = px.line(inversion,
              title="Evolución de una inversión de $1000 en cada acción", #calculamos la evolución de una inversión de $1000 en cada acción, dividiendo l
              labels={"value":"Valor de la inversión (USD)", "Date":"Fecha"},
              template="plotly_dark"#Grafica oscura, si queremos una grafica clara podemos poner "plotly_white"
              )
```

```python
fig.write_html("inversion_1000_dolares.html") # guardamos la grafica en un archivo html para poder compartirla con otras personas
```
