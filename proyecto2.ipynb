```python
#%pip install yfinance
#%pip install matplotlib
```

```python
import yfinance

tiker = input("Escriba el código de la acción que desea descargar: ")

data = yfinance.download(tiker, period="6mo")
data
```

```python
cierre = data.Close['AAPL']
cierre.plot(title="Precio de cierre de la acción AAPL en los últimos 6 meses", ylabel="Precio de cierre (USD)", xlabel="Fecha")
```

```python
maxima = round(cierre.max(), 2)
minima = round(cierre.min(), 2)
promedio = round(cierre.mean(), 2)

print(f"El precio máximo de cierre de la acción AAPL en los últimos 6 meses fue: {maxima} USD")
print(f"El precio mínimo de cierre de la acción AAPL en los últimos 6 meses fue: {minima} USD")
print(f"El precio promedio de cierre de la acción AAPL en los últimos 6 meses fue: {promedio} USD")
```

```python
import pyautogui
import pyperclip
import time
import webbrowser
```

```python
webbrowser.open("https://mail.google.com/mail/u/0/#inbox")
destinatario = "gcoscar170@gmail.com"
asunto = "Reporte de finanzas de los ultimo 6 meses"
mensaje = f'''
Buenas noches, este es el reporte de finanzas de los ultimos 6 meses de la empresa Apple.
El precio maximo de cierre es: {maxima} USD
El precio minimo de cierre es: {minima} USD
El precio promedio de cierre es: {promedio} USD
Pendiente de cualquier duda o comentario, quedo atento a sus comentarios.

oscar G
'''
# Espera 3 segundos para que cargue la página
time.sleep(3)
pyautogui.PAUSE = 3
# Click en el botón de redactar
pyautogui.click(x=418, y=398)

pyperclip.copy(destinatario)  # copia el correo del destinatario
pyautogui.hotkey("ctrl", "v")  # pega el correo del destinatario
pyautogui.press("tab")  # presiona tab para ir al siguiente campo

pyperclip.copy(asunto)  # copia el asunto del correo
pyautogui.hotkey("ctrl", "v")  # pega el asunto del correo
pyautogui.press("tab")  # presiona tab para ir al siguiente cpo

pyperclip.copy(mensaje)  # copia el mensaje del correo
pyautogui.hotkey("ctrl", "v")  # pega el mensaje del correo

pyautogui.click(x=1117, y=974)  # click en el botón de enviar
pyautogui.hotkey("ctrl", "w")  # cierra la pestaña del navegador




```
